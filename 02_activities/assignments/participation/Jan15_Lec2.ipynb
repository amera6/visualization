## Matplotlib

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate sample data
np.random.seed(613)  # For reproducibility
x=np.arange(50)
y=np.random.randint(1,100,50)
```

```python
# Create a scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
```

```python
# Create a bar plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
```

```python
# Create a line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
```

```python
# Activity (Slide 15 Lecture 2) - Create a histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
# Same plot as last time
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)

# Adding labels and title
ax.set_title("Total growth over time")
ax.set_ylabel("Total growth")
ax.set_xlabel("Years since start")
fig.tight_layout()
```

```python
# Creating dictionary for fonts
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
# Using dictionary on same line plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
# Customizing marker style
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker='*', color = "indigo")
```

```python
# Connect points with a line of specified colour, width, and style
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color = 'indigo', 
linestyle = '--', linewidth = 2)
```

```python
# Using hex code for colour
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
```

```python
# Customizing marker size and marker edge + face colours
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', 
    markersize = 12, 
    # color = '#7425b9', 
    linestyle = '--', 
    linewidth = 2, 
    markeredgecolor = '#fa9359', 
    markerfacecolor = '#000000' ) 
```

```python
# Add gridlines
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', 
    markersize = 12, 
    # color = '#7425b9', 
    linestyle = '--', 
    linewidth = 2, 
    markeredgecolor = '#fa9359', 
    markerfacecolor = '#000000' )
ax.grid(axis = 'y')
```

```python
# Activity (Slide 27 Lecture 2) - Customize appearance of gridlines
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
    markersize = 12, 
    linestyle = '--', 
    linewidth = 2, 
    markeredgecolor = '#fa9359', 
    markerfacecolor = '#000000' )
ax.grid(axis = 'y', color = 'gray', linestyle = '-.', linewidth = 0.5)
```

```python
# Activity (Slide 31 - Lecture 2) - Pick graph from Python Graph Gallery

# I chose Number 39 - Boxplot with Jitters

# Importing extra library
import seaborn as sns
```

```python
# Adding dataset:
a = pd.DataFrame({ 'group' : np.repeat('A',500), 'value': np.random.normal(10, 5, 500) })
b = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(13, 1.2, 500) })
c = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(18, 1.2, 500) })
d = pd.DataFrame({ 'group' : np.repeat('C',20), 'value': np.random.normal(25, 4, 20) })
e = pd.DataFrame({ 'group' : np.repeat('D',100), 'value': np.random.uniform(12, size=100) })
df = pd.concat([a,b,c,d,e])

# Making boxplot
ax = sns.boxplot(x='group', y='value', data=df)
# Add stripplot
ax = sns.stripplot(x='group', y='value', data=df, color="orange", jitter=0.2, size=2.5)

# Add title
plt.title("Boxplot with jitter", loc="left")

# Show the graph
plt.show()
```
