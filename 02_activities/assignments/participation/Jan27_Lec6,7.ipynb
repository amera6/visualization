```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

# Generate data
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Create subplots
fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
# Activity (slide 8) - Customize the plots
ax1.set_title("Scatter Plot of Random Data")
ax1.set_xlabel("X-axis Label")
ax1.set_ylabel("Y-axis Label")
ax2.set_title("Bar Chart of One Piece Characters")
ax2.set_xlabel("Characters")
ax2.set_ylabel("Power Levels")
fig.tight_layout()
```

```python
# Subplot mosaic example
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
# Add large axis titles - notice they don't fit well
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],['ax2', 'ax3']],figsize=(7, 4))

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

```python
# Improved layout with constrained option
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],['ax2', 'ax3']], figsize=(7, 4),
    layout = "constrained")
    
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# Multiple visualizations on same axes

# generate sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define figure and one axis
fig, ax = plt.subplots(figsize=(7, 3))

# add both bar and plot elements to the same axes
ax.bar(x, y1,
    color = "indigo")
ax.plot(x, y2,
    color = "red")
```

```python
# Adding error bars

# calculate standard deviations
y2_sd = np.std(y2)

# same line plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

# add error bars
ax.errorbar(x, 
    y2, 
    yerr = y2_sd,
    fmt = "none")
```

```python
# Customized error bars
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
    color = "red")
ax.errorbar(x,
    y2,
    yerr = y2_sd,
    fmt = "none",
    ecolor= "indigo",
    elinewidth= 4,
    capsize = 6,
    capthick= 4 )
```

```python
# Customize interval of error bars
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
    color = "red")

ax.errorbar(x,
    y2,
    yerr = y2_sd,
    fmt = "none",
    ecolor = "indigo",
    elinewidth = 4,
    capsize = 6,
    capthick = 4,
    errorevery= 2 )
```
