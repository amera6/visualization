```python
# Activity - Pick graph from gallery, recreate it, and comment on it

# I chose Number 39 Hidden Data Under Boxplot

# Importing extra library
import seaborn as sns
```

```python
# Adding dataset:
a = pd.DataFrame({ 'group' : np.repeat('A',500), 'value': np.random.normal(10, 5, 500) })
b = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(13, 1.2, 500) })
c = pd.DataFrame({ 'group' : np.repeat('B',500), 'value': np.random.normal(18, 1.2, 500) })
d = pd.DataFrame({ 'group' : np.repeat('C',20), 'value': np.random.normal(25, 4, 20) })
e = pd.DataFrame({ 'group' : np.repeat('D',100), 'value': np.random.uniform(12, size=100) })
df = pd.concat([a,b,c,d,e])

# Making boxplot
ax = sns.boxplot(x='group', y='value', data=df)
# Add stripplot
ax = sns.stripplot(x='group', y='value', data=df, color="orange", jitter=0.2, size=2.5)

# Add title
plt.title("Boxplot with jitter", loc="left")

# Show the graph
plt.show()
```

# Answer to Activity

In terms of Aesthetics, I personally believe the graph is pleasing to look at but that is highly subjective and dependent on the colour pallette used. The code from the gallery uses complementary colours of blue and orange which makes the jitters easier to view. 

Based on the description of the graph in the gallery, this graph appears to be substantive. It includes the basic boxplot which is useful for visualizing descriptive stats (like quartiles and outliers), but it also includes additional information about the underlying distribution for each boxplot and the number of observations (both of which are information that is lost in the basic boxplot). So, this visualization does accurately and honestly present the data. 

I think the one weakness of this visualization is its ability to convey its messgae to a lay audience. This percpetual gap comes from the fact that boxplots are already difficult to interpret unless you're well acquainted with statistical principles like distributions, skews, standard deviations, outliers, etc. Adding the dots on top could be informative to some viewers who desire a deeper understanding of the underlying distribution, but I believe that would be a smaller portion of the general audience who may misinterpret the jitters or find them to be overwhelming.
