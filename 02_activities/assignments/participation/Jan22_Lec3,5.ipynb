## Lecture 5 - Customizing Our Plots

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Generate sample data
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# Activity (Slide 5) - Line plot with two lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
```

```python
# Adding a legend to the plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='Line 1')
ax.plot(x, y2, label='Line 2')
ax.legend(loc='lower right')
```

```python
# Modifying the legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
    frameon = True, # add frame
    fontsize = 12, # font size
    ncol = 2, # number of columns
    shadow = True )
```

```python
# Move legend outside the plot area
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
    bbox_to_anchor =(1, 1))
```

```python
# Annotating plot points
# Create scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Specify coordinates and text
ax.text(10, 95, "This value is important!")
```

```python
# Modify text annotations

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
    ha='center', # alignment
    color = 'red', # font colour
    size = 20) # font size
```

```python
# Position text annotations

# Empty axes
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

# Add text in different coordinate systems
ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
# Annotating with arrows

# Create scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Add annotation with arrow
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(facecolor='black'))
```

```python
# Modifying arrows

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Add annotation with customized arrow
ax.annotate('This is important!',
    xy=(10, 95), xytext=(20, 94),
    arrowprops = dict(arrowstyle = "wedge", color = "hotpink"))
```

```python
# Removing tick marks and labels

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Use locator and formatter
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# Defining max number of ticks

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Use max number locator
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
# Setting tick intervals

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Use multiple locator
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# Rotating axis labels

# Same scatter plot as before
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

# Use rotation
plt.xticks(rotation=45, ha='right')
```

```python
# Activity (Slide 23) - Modifying axis title fonts
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.set_xlabel("Shiny New X Axis!", fontdict={'family':'serif','color':'indigo','size':14})
ax.legend(loc='lower right')
```
